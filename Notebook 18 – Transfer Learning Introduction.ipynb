### Simple Concept

**Transfer Learning** means using a model that has already learned from a large dataset and adapting it to a new, related task.

Instead of training a neural network from the beginning, we **reuse the knowledge learned by an existing model**.

### Simple Example

Suppose a pretrained model has already learned to recognize:

- Edges
- Shapes
- Textures
- Patterns
- Object parts

We can use this learned knowledge for a new image classification problem.

**Example:**

`Pretrained model → Cat/Dog dataset → Train for Cat vs Dog`

The model does not need to learn everything from zero.

### Basic Idea

**Without Transfer Learning:**

`Random Model → Train from scratch → Learn everything`

**With Transfer Learning:**

`Pretrained Model → Reuse learned knowledge → Adapt to new dataset`

Transfer learning usually requires **less training time and less data** than training a complete model from scratch.

### Real-World Example

A model trained on millions of general images may already understand basic visual patterns.

If we want to build a **plant disease classifier**, we can reuse that model and train it on plant images instead of building the entire vision model from zero.

### Important Point

Transfer learning is especially useful when:

- We have a **small or medium-sized dataset**
- We have **limited computing resources**
- A suitable pretrained model already exists
- The new task is related to what the pretrained model learned

**Main idea:**

> **Reuse existing knowledge instead of learning everything from scratch.**

### Simple Concept

**Feature extraction** means using the knowledge learned by a pretrained model to extract useful features from new images.

The pretrained model has already learned features such as:

- Edges
- Shapes
- Textures
- Patterns

We reuse these learned features instead of training those layers again.

### Why We Need Feature Extraction

We use feature extraction because **learning useful image features from scratch can require a lot of data and training time**.

A pretrained model has already learned many general visual patterns.

So instead of learning:

`Edges → Shapes → Textures → Patterns`

again, we can **reuse those learned features** and train only the part needed for our new task.

This is especially useful when:

- Our dataset is small.
- We have limited computing resources.

- The pretrained model is suitable for our image problem.

### How It Works

```text
New Image
   ↓
Pretrained Model
   ↓
Learned Features
   ↓
New Classification Layer
   ↓
Prediction
```

### Important Point

During feature extraction, the pretrained layers are usually **frozen**.

This means:

> The pretrained layers are used, but their weights are not updated during training.

Only the new task-specific layer is trained.


### Main Idea

**Feature Extraction = Reuse learned features from a pretrained model instead of learning them again from scratch.**

### Simple Concept

A **frozen layer** is a layer whose weights are **not updated during training**.

In transfer learning, we often freeze the layers of a pretrained model because those layers have already learned useful features.

For example:

```text
Pretrained Model

Layer 1 → Frozen
Layer 2 → Frozen
Layer 3 → Frozen
Layer 4 → Frozen
Output Layer → Trainable
```

The frozen layers continue to extract the features they already learned.

### Why Do We Need Frozen Layers?

We freeze layers because:

- We want to **preserve the pretrained knowledge**.
- We don't need to retrain all layers.
- Training becomes faster.
- Fewer parameters need to be updated.
- It is useful when our new dataset is small.



### Main Idea

**Frozen layer = Use the learned knowledge, but don't change its weights.**

````python
### Simple Python Example

from torchvision import models

# Load pretrained model
model = models.resnet18(weights="DEFAULT")

# Freeze all layers
for parameter in model.parameters():
    parameter.requires_grad = False

# Check whether parameters are frozen
print(model.conv1.weight.requires_grad)
````

### Simple Concept

A **trainable layer** is a layer whose weights **can be updated during training**.

In transfer learning, we usually keep most pretrained layers frozen and make the new task-specific layer trainable.

```text
Pretrained Model

Feature Layer 1 → Frozen
Feature Layer 2 → Frozen
Feature Layer 3 → Frozen
Output Layer    → Trainable
```

The trainable layer learns how to use the extracted features for our **new task**.

### Why Do We Need Trainable Layers?

The pretrained model was trained for a different task.

For example, a pretrained model may originally classify **1000 different image categories**.

Our new problem might have only:

`Cat vs Dog`

Therefore, we need a new output layer that can learn our specific classes.

### Main Idea

**Frozen layers → Reuse existing knowledge**

**Trainable layers → Learn the new task**

````python

### Simple Python Example

from torchvision import models
import torch.nn as nn

# Load pretrained model
model = models.resnet18(weights="DEFAULT")

# Freeze pretrained layers
for parameter in model.parameters():
    parameter.requires_grad = False

# Replace the final layer
model.fc = nn.Linear(model.fc.in_features, 2)

# Check the new layer
print(model.fc.weight.requires_grad)
````

### Simple Concept

**Fine-tuning** means taking a pretrained model and **training some of its pretrained layers again** on our new dataset.

Unlike feature extraction, where pretrained layers remain frozen, fine-tuning allows selected layers to become trainable.

### Why Do We Need Fine-Tuning?

A pretrained model has learned general image features, but our new dataset may have **specific patterns** that the original model did not learn.

Fine-tuning helps the model **adapt its existing knowledge to the new task**.

### Feature Extraction vs Fine-Tuning

```text
Feature Extraction
→ Pretrained layers stay frozen
→ Train mainly the new output layer

Fine-Tuning
→ Some pretrained layers are unfrozen
→ Train those layers together with the new output layer
```


### Main Idea

**Fine-tuning = Reuse a pretrained model and allow some of its layers to learn from the new dataset.**

````python

### Simple Python Example

from torchvision import models

# Load pretrained model
model = models.resnet18(weights="DEFAULT")

# Freeze all layers
for parameter in model.parameters():
    parameter.requires_grad = False

# Unfreeze the final layer
for parameter in model.layer4.parameters():
    parameter.requires_grad = True

print(model.layer4[0].conv1.weight.requires_grad)
````

### Simple Concept

The general transfer learning process is:

```text
Pretrained Model
       ↓
Freeze / Modify Layers
       ↓
Train on New Dataset
       ↓
Fine-Tune
       ↓
Final Model
```

### Step 1 – Load a Pretrained Model

Start with an existing model such as **ResNet-18**.

### Step 2 – Freeze Layers

Freeze the pretrained layers so their learned features are preserved.

### Step 3 – Modify the Output Layer

Replace the original output layer according to the number of classes in the new dataset.

### Step 4 – Train on the New Dataset

Train the new output layer using your dataset.

### Step 5 – Fine-Tune

If needed, unfreeze some pretrained layers and train them with a **small learning rate** so the model can adapt better to the new task.

### Simple Workflow Example

```text
ResNet-18 Pretrained
        ↓
Freeze Feature Layers
        ↓
Replace Output Layer
        ↓
Train New Output Layer
        ↓
Unfreeze Selected Layers
        ↓
Fine-Tune
        ↓
Final Model
```

### Main Idea

Transfer learning is not simply **"use a pretrained model."**

It is a process of:

**Reuse → Modify → Train → Fine-Tune**

### Simple Concept

Now we can combine the main transfer learning steps into a simple PyTorch example.

We will use **ResNet-18**, load its pretrained weights, freeze the pretrained layers, and replace the final layer for a **2-class classification problem**.


### What the Code Does

- **Load pretrained model** → gets ResNet-18 with learned weights.
- **Freeze layers** → prevents the pretrained layers from changing.
- **Replace `fc`** → creates a new output layer for 2 classes.
- **Check parameters** → shows that only the new `fc` layer is trainable.

### Complete Workflow

```text id="9b6d1r"
Pretrained ResNet-18
        ↓
Freeze Layers
        ↓
Replace Output Layer
        ↓
Train New Layer
        ↓
Optional Fine-Tuning
        ↓
Final Model
```

### Main Idea

This is the basic implementation pattern you will commonly see when using **transfer learning with PyTorch/TorchVision**.

````python
### Python Implementation

import torch
import torch.nn as nn
from torchvision import models

# 1. Load pretrained ResNet-18
model = models.resnet18(weights="DEFAULT")

# 2. Freeze pretrained layers
for parameter in model.parameters():
    parameter.requires_grad = False

# 3. Replace the final layer
model.fc = nn.Linear(model.fc.in_features, 2)

# 4. Check trainable parameters
for name, parameter in model.named_parameters():
    if parameter.requires_grad:
        print(name)
````
