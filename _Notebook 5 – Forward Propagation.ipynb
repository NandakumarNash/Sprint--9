# Notebook 5 – Forward Propagation

## 1. Forward Propagation

### 1. Simple Concept

**Forward propagation** means sending input data through a neural network from **start to finish** to get a prediction.

The flow is:

```text
Input
  ↓
Hidden Layer
  ↓
Output Layer
  ↓
Prediction
```

For example, suppose we want to predict whether a customer will buy a product.

The input could be:

```text
Age = 30
Monthly Income = 50000
```

The neural network takes these inputs and processes them through its layers.

Each layer performs some calculations and passes the result to the next layer.

Finally, the output layer produces the prediction.

So:

```text
Customer Information
        ↓
Input Layer
        ↓
Hidden Layer
        ↓
Output Layer
        ↓
Purchase Prediction
```



### 4. Concept-Based Real-World Example

Imagine an online shopping company wants to predict:

> **Will this customer buy a laptop?**

The customer information is:

```text
Age = 30
Monthly Income = ₹50,000
Previous Purchases = 5
```

These values enter the neural network.

```text
Age
Monthly Income
Previous Purchases
        ↓
   Input Layer
        ↓
   Hidden Layer
        ↓
   Output Layer
        ↓
   Buy = 1
```

The important idea is:

> **Forward propagation is the complete journey of input data through the neural network until a prediction is produced.**

We will now learn what happens to the input inside each layer.

# 3. Hidden Layer

## 1. Simple Concept

A **hidden layer** is the layer between the input layer and the output layer.

Its job is to **process the input data** and find useful patterns.

Example:

```text
Input Layer
    ↓
Hidden Layer
    ↓
Output Layer
```

A hidden layer contains multiple **neurons**.

For example:

```text
Age ──────────────→ ● Neuron 1
                    ● Neuron 2
Income ───────────→ ● Neuron 3
                    ● Neuron 4
Purchases ────────→
```

Each neuron receives the input values and performs calculations using:

* **Weights**
* **Bias**
* **Activation function**

The result is then passed to the next layer.

### Why is it called "hidden"?

Because we don't directly provide the hidden-layer values as input, and we don't directly see them as the final answer.

The neural network **learns these values internally**.

---

---

A hidden neuron might learn a pattern such as:

> "Customers with higher income and more previous purchases may be more likely to buy."

The neuron combines the inputs using weights and bias:

```text
Age ────────────────┐
                    │
Income ─────────────┼──→ Hidden Neuron
                    │
Purchases ──────────┘
                         ↓

Multiple hidden neurons can learn **different patterns** from the same customer data.

For example:

```text
Hidden Neuron 1 → Income-related pattern
Hidden Neuron 2 → Purchase-history pattern
Hidden Neuron 3 → Age-related pattern
```

These processed values are then passed to the **output layer**, which eventually produces the prediction.

### Key idea

> **The hidden layer takes the input, combines it using weights and bias, and creates new values that help the network find patterns.**


````python

## 2. Simple Python Example

#This topic requires implementation, so let's create a very small hidden layer using NumPy.

import numpy as np

# Input data
customer_data = np.array([30, 50000, 5])

# Weights for one hidden neuron
weights = np.array([0.2, 0.0001, 0.5])

# Bias
bias = 1

# Weighted sum
z = np.dot(customer_data, weights) + bias

print("Hidden neuron value:", z)

````

# 4. Weights and Biases

## 1. Simple Concept

A **weight** tells a neuron how much importance to give to an input.

A **bias** is an extra value that helps the neuron adjust its calculation.

Think of a neuron like a small calculator:

```text
Input
  ↓
Weight
  ↓
Calculation
  ↓
Bias added
  ↓
Result
```

For example:

```text
Age                = 30
Monthly Income     = 50,000
Previous Purchases = 5
```

The neuron may have:

```text
Age weight                = 0.2
Income weight             = 0.0001
Purchases weight          = 0.5
Bias                      = 1
```

The neuron calculates:

```text
(30 × 0.2)
+
(50,000 × 0.0001)
+
(5 × 0.5)
+
1
```

The result is:

```text
14.5
```

Remember:

> **14.5 is not "good" or "bad." It is simply the result produced by this neuron.**

The next part of the neural network will use this result.

---

## 4. Concept-Based Real-World Example

Imagine an e-commerce company wants to predict whether a customer will buy a laptop.

The network receives:

```text
Age                = 30
Monthly Income     = ₹50,000
Previous Purchases = 5
```

The neuron gives different weights to these inputs:

```text
Age                → weight = 0.2
Monthly Income     → weight = 0.0001
Previous Purchases → weight = 0.5
```

Why different weights?

Because the network may learn that different inputs contribute differently to the prediction.

The calculation becomes:

```text
Age contribution
30 × 0.2 = 6

Income contribution
50,000 × 0.0001 = 5

Purchase contribution
5 × 0.5 = 2.5

Bias
= 1
```

Then:

```text
6 + 5 + 2.5 + 1 = 14.5
```

So the hidden neuron produces:

```text
14.5
```

This value is then passed forward.

### Key idea

> **Weights control how strongly each input contributes, while bias gives the neuron an additional adjustment.**

And the important flow is:

```text
Input
  ↓
Multiply by weights
  ↓
Add everything
  ↓
Add bias
  ↓
Neuron value
```


````python

## 2. Simple Python Example

import numpy as np

# Input values
age = 30
monthly_income = 50000
previous_purchases = 5

# Weights
age_weight = 0.2
income_weight = 0.0001
purchase_weight = 0.5

# Bias
bias = 1

# Calculate weighted inputs
age_part = age * age_weight
income_part = monthly_income * income_weight
purchase_part = previous_purchases * purchase_weight

# Calculate neuron value
z = age_part + income_part + purchase_part + bias

print("Age contribution:", age_part)
print("Income contribution:", income_part)
print("Purchase contribution:", purchase_part)
print("Bias:", bias)
print("Neuron value:", z)
````

# 5. Linear Transformation

## 1. Simple Concept

**Linear transformation** is the calculation a neuron performs using:

* Inputs
* Weights
* Bias

In simple words:

> **The neuron takes the input values, multiplies them by their weights, adds them together, and then adds the bias.**

The basic calculation is:

```text
z = (input × weight) + (input × weight) + bias
```

For multiple inputs:

```text
z = x₁w₁ + x₂w₂ + x₃w₃ + b
```

Using our customer example:

```text
Age                = 30
Monthly Income     = 50,000
Previous Purchases = 5
```

Weights:

```text
Age weight         = 0.2
Income weight      = 0.0001
Purchase weight    = 0.5
```

Bias:

```text
b = 1
```

Calculation:

```text
z = (30 × 0.2)
  + (50000 × 0.0001)
  + (5 × 0.5)
  + 1

z = 6 + 5 + 2.5 + 1

z = 14.5
```

So the **linear transformation produces 14.5**.

Remember:

> **14.5 is just the result of the calculation. It is not yet a prediction.**


## 4. Concept-Based Real-World Example

Imagine a bank wants to predict whether a customer may take a loan.

The inputs could be:

```text

Existing Loans
```

The neuron assigns weights to these inputs.

For example:

```text
Monthly Income → weight
Credit Score   → weight
Existing Loans → weight
Bias           → adjustment
```

The neuron combines them:

```text
Inputs
  ↓
Multiply each input by its weight
  ↓
Add the results
  ↓
Add bias
  ↓
Linear transformation result
```

This result is then sent to an **activation function**.

So the complete flow is:

```text
Input
   ↓
Weight × Input
   ↓
Add everything
   ↓
Add Bias
   ↓
Linear Transformation
   ↓
Activation
   ↓
Next Layer
```

### Key idea

> **Linear transformation is the basic calculation performed by a neuron before applying an activation function.**


````python

## 2. Simple Python Example

import numpy as np

# Input values
customer_data = np.array([30, 50000, 5])

# Weights
weights = np.array([0.2, 0.0001, 0.5])

# Bias
bias = 1

# Linear transformation
z = np.dot(customer_data, weights) + bias

print("Linear transformation result:", z)
````

# 6. Activation Function

## 1. Simple Concept

After a neuron performs the linear calculation, we get a value such as:

```text
z = 14.5
```

The **activation function** takes this value and decides what value should be passed to the next layer.

So:

```text
Input
  ↓
Weights + Bias
  ↓
Linear Transformation
  ↓
Activation Function
  ↓
Next Layer
```

### Why do we need activation?

Without activation functions, a neural network would mostly behave like one large linear calculation.

Activation functions allow the network to learn **complex patterns**.

For our hidden layer, we can use **ReLU**.

ReLU means:

> **If the value is positive, keep it. If it is negative, change it to 0.**

```text
ReLU(x) = max(0, x)
```

Examples:

```text
ReLU(14.5) = 14.5
ReLU(-3)   = 0
ReLU(0)    = 0
```

So if our neuron produced:

```text
14.5
```

ReLU gives:

```text
14.5
```

The important point is:

> **Activation does not necessarily mean "good" or "bad." It transforms the neuron's value before passing it forward.**

---

## 4. Concept-Based Real-World Example

Imagine a neural network is processing customer information.

The hidden neuron calculates:

```text
Linear transformation
        ↓
       14.5
```

Then ReLU is applied:

```text
14.5
 ↓
ReLU
 ↓
14.5
```

This processed value is sent to the next layer.

If another neuron produces:

```text
-4
```

then:

```text
-4
 ↓
ReLU
 ↓
0
```

So activation helps the neural network decide **what signal to pass forward**.

The complete process is now:

```text
Customer Data
     ↓
Weights + Bias
     ↓
Linear Transformation
     ↓
14.5
     ↓
ReLU
     ↓
14.5
     ↓
Next Layer
```

### Key idea

> **Linear transformation calculates a value. Activation transforms that value before sending it to the next layer.**

This is why our forward propagation flow is:

```text
Input
 ↓
Linear Transformation
 ↓
Activation
 ↓
Next Layer
```


````python

## 2. Simple Python Example

import numpy as np

# Linear transformation result
z = 14.5

# ReLU activation
activated_value = np.maximum(0, z)

print("Before activation:", z)
print("After ReLU:", activated_value)

````

# 7. Output Layer

## 1. Simple Concept

The **output layer** is the final layer of the neural network.

Its job is to take the information coming from the hidden layer and produce the **final output**.

Our flow is now:

```text
Input Layer
     ↓
Hidden Layer
     ↓
Output Layer
     ↓
Prediction
```

The hidden layer produces processed values.

The output layer uses those values to calculate the final result.

For example:

```text
Hidden values
     ↓
Weights + Bias
     ↓
Linear calculation
     ↓
Output
```

For a **binary classification** problem, such as:

> Will the customer buy the laptop?

the output is usually converted into a probability between `0` and `1`.

For example:

```text
0.82
```

This can be interpreted as:

> The model estimates an **82% probability** of the customer belonging to the "buy" class.

---

---

## 4. Concept-Based Real-World Example

Suppose we are predicting whether a customer will buy a laptop.

The hidden layer has already processed the customer's information:

```text

```text

### Key idea

> **The output layer takes the processed information from the hidden layer and produces the final value used for prediction.**


````python

## 2. Simple Python Example

#Let's say our hidden layer has produced two values:


#Hidden value 1 = 14.5
#Hidden value 2 = 8.2


#The output neuron uses them:


import numpy as np

# Values coming from hidden layer
hidden_values = np.array([14.5, 8.2])

# Output layer weights
output_weights = np.array([0.1, 0.2])

# Output bias
output_bias = 0.5

# Linear calculation in output layer
z_output = np.dot(hidden_values, output_weights) + output_bias

print("Output layer value:", z_output)

````

> **Input values are multiplied by weights, bias is added, activation is applied, the result moves to the next layer, and finally the output layer produces a value that can be converted into a prediction.**

That entire journey is **forward propagation**.


````python

## 2. Simple Python Example

import numpy as np

# -----------------------------
# Step 1: Input
# -----------------------------
X = np.array([30, 50000])

# -----------------------------
# Step 2: Input → Hidden Layer
# -----------------------------

# Weights
W_hidden = np.array([
    [0.2, 0.1],
    [0.0001, 0.0002]
])

# Bias
b_hidden = np.array([1, 0.5])

# Linear transformation
z_hidden = np.dot(X, W_hidden) + b_hidden

print("Hidden layer before activation:", z_hidden)

# -----------------------------
# Step 3: Activation
# -----------------------------

# ReLU
a_hidden = np.maximum(0, z_hidden)

print("Hidden layer after ReLU:", a_hidden)

# -----------------------------
# Step 4: Hidden → Output Layer
# -----------------------------

# Output weights
W_output = np.array([0.1, 0.2])

# Output bias
b_output = 0.5

# Output linear transformation
z_output = np.dot(a_hidden, W_output) + b_output

print("Output before Sigmoid:", z_output)

# -----------------------------
# Step 5: Sigmoid
# -----------------------------

probability = 1 / (1 + np.exp(-z_output))

print("Prediction probability:", probability)
````

# 10. Forward Propagation Using PyTorch

## 1. Simple Concept

In the previous section, we manually performed:

```text
Input
 ↓
Hidden Layer
 ↓
ReLU
 ↓
Output Layer
 ↓
Sigmoid
 ↓
Prediction
```

Using NumPy, we calculated each step ourselves.

With **PyTorch**, we can create the neural network layers and let PyTorch perform these calculations.

Our network is:

```text
2 Inputs
   ↓
2 Hidden Neurons
   ↓
ReLU
   ↓
1 Output Neuron
   ↓
Sigmoid
   ↓
Prediction Probability
```

The important point is:

> **PyTorch is doing the same forward-propagation process, but the framework handles the calculations for us.**

---


## 4. Concept-Based Real-World Example

Imagine an e-commerce company wants to predict whether a customer will buy a laptop.

The input is:

```text
Age = 30
Monthly Income = ₹50,000
```

PyTorch receives:

```text
[30, 50000]
```

### Step 1 — Hidden layer

```text
2 inputs
   ↓
2 hidden neurons
```

Each hidden neuron performs its own weighted calculation.

### Step 2 — ReLU

The hidden values pass through ReLU.

```text
Hidden values
     ↓
   ReLU
     ↓
Processed hidden values
```

### Step 3 — Output layer

The two hidden values are combined by the output neuron.

```text
2 hidden values
       ↓
1 output neuron
       ↓
Raw output
```

### Step 4 — Sigmoid

The raw output is converted into a probability.

```text
Raw output
    ↓
 Sigmoid
    ↓
  0.82
```

### Step 5 — Prediction

```text
0.82 >= 0.5
```

Therefore:

```text
Prediction = 1
```

Meaning:

```text
Customer → Predicted to buy
```

### ⭐ The complete idea

```text
Customer Data
     ↓
[30, 50000]
     ↓
Input Layer
     ↓
Linear Transformation
     ↓
Hidden Neurons
     ↓
ReLU
     ↓
Output Neuron
     ↓
Sigmoid
     ↓
0.82 probability
     ↓
Prediction = 1
```

> **This is how an input finally becomes a prediction through forward propagation.**


````python

## 2. Simple Python Example

import torch
import torch.nn as nn

# Input
X = torch.tensor([[30.0, 50000.0]])

# Create the network
model = nn.Sequential(
    nn.Linear(2, 2),
    nn.ReLU(),
    nn.Linear(2, 1),
    nn.Sigmoid()
)

# Forward propagation
output = model(X)

print("Prediction probability:", output.item())
````
