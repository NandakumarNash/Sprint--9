# 1. Overfitting

## Simple Concept

**Overfitting** happens when a neural network learns the training data **too well**, including its noise and unnecessary patterns.

As a result:

- Training performance → **very good**
- Validation/test performance → **poor**

### Simple Example

Suppose a model learns to identify cats.

It performs very well on the training images because it remembers specific details from those images.

But when it sees **new cat images**, its performance becomes poor.

That is overfitting.

## How to Identify Overfitting

Usually:

```text
Training Loss    ↓ keeps decreasing
Validation Loss  ↑ starts increasing
```

or:

```text
Training Accuracy    ↑ very high
Validation Accuracy  → stops improving / decreases
```

## Why Does Overfitting Happen?

Common reasons:

- Model is too complex
- Too many training epochs
- Training dataset is too small
- Model learns noise instead of useful patterns

## Important Point

The goal of a neural network is **not just to perform well on training data**.

The goal is to **generalize well to unseen data**.

```text
Good Model
Training Data → learns useful patterns
New Data      → performs well
```

```text
Overfit Model
Training Data → performs very well
New Data      → performs poorly
```

## Key Point

**Overfitting = Good training performance + Poor validation/test performance.**

# 2. Underfitting

## Simple Concept

**Underfitting** happens when a neural network is **too simple or has not learned enough** from the training data.

As a result:

- Training performance → **poor**
- Validation/test performance → **also poor**

### Simple Example

Suppose we use a very simple model to predict house prices.

The model does not learn important patterns such as:

- House size
- Number of bedrooms
- Location

So it performs poorly on both the training data and new data.

That is underfitting.

## How to Identify Underfitting

Usually:

```text
Training Loss    → High
Validation Loss  → High
```

and:

```text
Training Accuracy    → Low
Validation Accuracy  → Low
```

## Why Does Underfitting Happen?

Common reasons:

- Model is too simple
- Too few training epochs
- Model has insufficient layers/neurons
- Important features are missing
- Learning has not progressed enough

## Overfitting vs Underfitting

| | Underfitting | Overfitting |
|---|---|---|
| Training performance | Poor | Very good |
| Validation performance | Poor | Poor |
| Main problem | Model learns too little | Model learns too much/noise |

## Key Point

**Underfitting = Poor training performance + Poor validation/test performance.**

# 3. Bias

## Simple Concept

**Bias** is the error caused when a model is **too simple** and cannot capture the important patterns in the data.

### Simple Example

Imagine a neural network trying to predict house prices.

If the model is too simple, it may not properly learn the relationship between:

- House size
- Number of bedrooms
- Location
- House price

The model makes similar mistakes even on the training data.

This is called **high bias**.

## High Bias

```text
High Bias
    ↓
Model is too simple
    ↓
Does not learn enough
    ↓
Training performance is poor
    ↓
Underfitting
```

## Important Point

Bias is mainly associated with **underfitting**.

```text
High Bias → Underfitting
```

## Key Point

**Bias = Error caused because the model is too simple to learn the important patterns.**

# 4. Variance

## Simple Concept

**Variance** describes how much a model's predictions can change when the training data changes.

In Deep Learning, **high variance** usually means the model is learning the training data too specifically.

This can lead to **overfitting**.

### High Variance

```text
High Variance
      ↓
Model is very sensitive to training data
      ↓
Learns specific/noisy patterns
      ↓
Training performance becomes very good
      ↓
Performance on new data becomes poor
      ↓
Overfitting
```

### Low Variance

A model with lower variance is less sensitive to small changes in the training data and can generalize better.

## Bias vs Variance

```text
High Bias     → Model too simple → Underfitting
High Variance → Model too complex → Overfitting
```



## Key Point

**Variance = How sensitive the model is to changes in the training data.**

**High variance is commonly associated with overfitting.**

# 5. Generalization

## Simple Concept

**Generalization** means a model can perform well on **new, unseen data**, not just on the data it was trained on.

### Simple Example

Suppose we train a neural network using 1,000 customer records.

After training, we give it 200 new customer records that it has never seen before.

If the model performs well on those new records, it has **good generalization**.

## Good Generalization

```text
Training Data
     ↓
Model learns useful patterns
     ↓
New / Unseen Data
     ↓
Good Predictions
```

## Poor Generalization

If the model performs very well on training data but poorly on new data, it has **poor generalization**.

This commonly happens because of **overfitting**.

## Why Generalization Is Important

In real-world AI/ML applications, the model will usually receive **new data** after deployment.

Therefore, we want the model to learn **general patterns**, rather than simply memorize the training data.

## Key Point

**Generalization = How well a model performs on unseen/new data.**

A good deep learning model should have **good generalization**.

# 6. Dropout

## Simple Concept

**Dropout** is a regularization technique used to reduce **overfitting** in neural networks.

During training, Dropout randomly turns off some neurons.

```text
Without Dropout
Input → Neuron → Neuron → Neuron → Output
              All neurons active
```

```text
With Dropout
Input → Neuron → X → Neuron → Output
              Some neurons temporarily turned off
```

The neurons that are turned off are selected **randomly** during each training step.

## Why Use Dropout?

Without Dropout, a neural network may depend too much on particular neurons and memorize the training data.

Dropout forces the network to learn using different combinations of neurons.

This can help the model **generalize better** to unseen data.

## Important Point

Dropout is mainly applied **during training**.

During evaluation/testing, Dropout is automatically disabled, so the full network is used.

## Example

```text
nn.Dropout(0.5)
```

`0.5` means approximately **50% of the neurons are randomly dropped during training**.

## Key Point

**Dropout = Randomly turn off some neurons during training to reduce overfitting.**

````python
# Python Example — Dropout

import torch
import torch.nn as nn

# Create Dropout layer
dropout = nn.Dropout(p=0.5)

# Input values
x = torch.tensor([1.0, 2.0, 3.0, 4.0])

# Apply dropout
output = dropout(x)

print("Input :", x)
print("Output:", output)
````

```

### What this code does

- `nn.Dropout(p=0.5)` → randomly drops about **50%** of values during training.
- `x` → input values.
- `dropout(x)` → applies Dropout to the input.
- Some values become `0`.

The exact output can change each time because Dropout is **random**.

For example:

```text
Input :  tensor([1., 2., 3., 4.])
Output: tensor([0., 4., 0., 8.])
```

The remaining values are scaled during training so that the overall expected value is maintained.

# 7. L1 Regularization

## Simple Concept

**L1 regularization** adds a penalty to the model when its weights become large.

It encourages some weights to become **zero or very close to zero**.

```text
Large weights
     ↓
L1 penalty
     ↓
Weights become smaller
     ↓
Some weights may become 0
     ↓
Simpler model
```

## Why Use L1 Regularization?

L1 regularization can help reduce **overfitting** by making the model simpler.

It can also make the model focus on the **more important features** because some less useful weights may become zero.

## Important Point

L1 regularization adds the **absolute values of weights** as a penalty to the loss.

```text
Total Loss = Original Loss + L1 Penalty
```

## Key Point

**L1 regularization = Penalizes large weights and can make some weights zero, helping reduce overfitting.**

````python
# Python Example — L1 Regularization

import torch
import torch.nn as nn

# Create a simple model
model = nn.Linear(3, 1)

# L1 penalty
l1_penalty = torch.sum(torch.abs(model.weight))

print("L1 Penalty:", l1_penalty.item())
````

```

### What this code does

- `model.weight` → gets the model's weights.
- `torch.abs()` → converts weights to their absolute values.
- `torch.sum()` → adds all absolute weight values.
- The result is the **L1 penalty**.

In actual model training, this penalty is added to the normal loss:

```python
loss = original_loss + lambda_value * l1_penalty
```

`lambda_value` controls **how strongly L1 regularization is applied**.

# 8. L2 Regularization

## Simple Concept

**L2 regularization** adds a penalty when the model's weights become large.

It encourages the model to keep weights **small**, which can help reduce overfitting.

```text
Large weights
     ↓
L2 penalty
     ↓
Weights become smaller
     ↓
Model becomes less sensitive
     ↓
Better generalization
```

## Why Use L2 Regularization?

A neural network with very large weights can become too focused on specific training patterns.

L2 regularization discourages very large weights and helps the model learn smoother, more general patterns.

## Important Point

L2 regularization adds the **squared values of the weights** as a penalty.

```text
Total Loss = Original Loss + L2 Penalty
```

Unlike L1 regularization, L2 usually makes weights **smaller**, rather than making many weights exactly zero.

## L1 vs L2

| L1 | L2 |
|---|---|
| Uses absolute weights | Uses squared weights |
| Can make weights zero | Usually makes weights small |
| Can produce sparse models | Produces smoother models |

## Key Point

**L2 regularization = Penalizes large weights and helps reduce overfitting.**

````python
# Python Example — L2 Regularization

import torch
import torch.nn as nn

# Create a simple model
model = nn.Linear(3, 1)

# L2 penalty
l2_penalty = torch.sum(model.weight ** 2)

print("L2 Penalty:", l2_penalty.item())
````

```

### What this code does

- `model.weight` → gets the model's weights.
- `model.weight ** 2` → squares each weight.
- `torch.sum()` → adds the squared weights.
- The result is the **L2 penalty**.

During training, the penalty can be added to the original loss:

```python
loss = original_loss + lambda_value * l2_penalty
```

`lambda_value` controls how strongly L2 regularization is applied.

# 9. Weight Decay

## Simple Concept

**Weight decay** is a technique that encourages neural network weights to remain **small** during training.

It is closely related to **L2 regularization**.

```text
Large weights
     ↓
Weight decay
     ↓
Weights gradually become smaller
     ↓
Less overfitting
```

## Why Use Weight Decay?

Very large weights can make a model depend too strongly on particular training patterns.

Weight decay discourages large weights and can help the model **generalize better** to unseen data.

## Python Example

In PyTorch, weight decay can be added directly to the optimizer:

```python
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001,
    weight_decay=0.01
)
```

Here:

- `lr=0.001` → learning rate
- `weight_decay=0.01` → applies weight decay

## L2 vs Weight Decay

For basic learning, you can remember:

```text
L2 Regularization → adds a penalty for large weights
Weight Decay      → directly encourages weights to become smaller
```

They are closely related, and in many common optimizer setups, **weight decay is used as the practical way to apply L2-style regularization**.

## Key Point

**Weight decay = Encourages smaller weights during training to help reduce overfitting.**

# 10. Batch Normalization

## Simple Concept

**Batch Normalization (BatchNorm)** normalizes the values flowing through a neural network layer during training.

It helps keep the values in a more stable range while the network is learning.

```text id="8w8l1p"
Input
  ↓
Linear Layer
  ↓
Batch Normalization
  ↓
Activation
  ↓
Next Layer
```

## Why Use Batch Normalization?

Batch Normalization can:

- Make training more stable
- Help the model train faster
- Sometimes reduce overfitting
- Improve generalization in some situations

## Python Example

```python id="s8j3qf"
import torch.nn as nn

batch_norm = nn.BatchNorm1d(10)

print(batch_norm)
```

Here:

- `10` → number of features coming into the BatchNorm layer.
- `BatchNorm1d` → commonly used for tabular data and fully connected neural networks.

## Important Point

Batch Normalization is **not primarily an overfitting technique** like Dropout.

Its main purpose is to **normalize activations and stabilize training**, although it can sometimes provide a regularization effect.

## Key Point

**Batch Normalization = Normalizes layer outputs to make neural network training more stable.**

** Real-World Example — Batch Normalization**
Imagine a neural network predicting house prices.
The input features may have very different scales:
House Size     → 1500
Bedrooms       → 3
House Age      → 10

After passing through neural network layers, the values can also become very different.
Batch Normalization helps keep these intermediate values more stable while the network is training.

# 11. Early Stopping

## Simple Concept

**Early stopping** stops neural network training when the model stops improving on the **validation data**.

The main purpose is to prevent **overfitting**.

### Example

Suppose we train a model for 20 epochs:

```text
Epoch     Training Loss     Validation Loss
  1           0.80                0.85
  5           0.50                0.55
 10           0.30                0.40
 15           0.15                0.60
 20           0.08                0.90
```

After around epoch 10:

- Training loss continues decreasing.
- Validation loss starts increasing.

This indicates that the model may be **overfitting**.

Early stopping can stop training around the point where validation performance was best.

## Important Term — Patience

**Patience** means how many epochs we allow the validation performance to stop improving before stopping training.

Example:

```text
patience = 3
```

means:

> Allow 3 more epochs without improvement before stopping.

## Why Use Early Stopping?

```text
Training
   ↓
Validation performance improves
   ↓
Continue training
   ↓
Validation performance stops improving
   ↓
Wait for patience period
   ↓
Stop training
```

## Key Point

**Early stopping = Stop training when validation performance stops improving, helping prevent overfitting.**

# 13. Training and Validation Curves

## Simple Concept

A **training curve** shows how the model performs on the training data during each epoch.

A **validation curve** shows how the model performs on unseen validation data during each epoch.

These curves help us identify **overfitting and underfitting**.

## Loss Curves

We usually plot:

```text
Training Loss
Validation Loss
```

### Good Training

Both losses decrease and remain reasonably close.

```text
Training Loss    ↓
Validation Loss  ↓
```

This suggests the model is learning and generalizing reasonably well.

### Overfitting

```text
Training Loss    ↓↓↓
Validation Loss  ↓ then ↑
```

Training loss continues to improve, but validation loss starts increasing.

This is a common sign of **overfitting**.

## Accuracy Curves

For classification, we can also plot:

```text
Training Accuracy
Validation Accuracy
```

If training accuracy keeps increasing while validation accuracy stops improving or decreases, the model may be overfitting.

## In Our Experiment

We will compare the curves of:

```text
Without Dropout
        VS
With Dropout
```

The goal is to check whether Dropout produces better validation performance and reduces the gap between training and validation performance.

## Key Point

**Training and validation curves help us understand how well the model learns and whether it is overfitting.**

````python
# Plot Training and Validation Curves

import matplotlib.pyplot as plt

# Example training history
epochs = range(1, 11)

train_loss = [0.80, 0.65, 0.52, 0.42, 0.34, 0.28, 0.23, 0.19, 0.16, 0.13]
val_loss = [0.82, 0.68, 0.55, 0.47, 0.42, 0.40, 0.42, 0.46, 0.51, 0.58]

plt.plot(epochs, train_loss, label="Training Loss")
plt.plot(epochs, val_loss, label="Validation Loss")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training vs Validation Loss")
plt.legend()
plt.show()
````

```

### What the graph tells us

- **Training loss decreases** → model is learning the training data.
- **Validation loss decreases initially** → model is learning useful patterns.
- **Validation loss starts increasing** while training loss continues decreasing → possible **overfitting**.

In the example, validation loss starts increasing after around **epoch 6**, so the model may be starting to overfit.
