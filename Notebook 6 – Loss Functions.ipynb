# 1. Why Do We Need a Loss Function?

## Simple Concept

A **loss function tells us how wrong the model's prediction is**.

The model makes a prediction:

**Input → Model → Prediction**

Then we compare:

**Prediction vs Actual Answer**

The loss function gives a number showing the error.

### Simple Example

Suppose we want to predict house price.

Actual price:

```text
₹50 Lakhs
```

Model prediction:

```text
₹45 Lakhs
```

The prediction is wrong by ₹5 Lakhs.

The **loss function converts this error into a numerical value**.

---

## Prediction → Loss → Learning

The basic Deep Learning process is:

```text
Input
  ↓
Neural Network
  ↓
Prediction
  ↓
Compare with Actual Value
  ↓
Loss Function
  ↓
Loss Value
  ↓
Model adjusts weights
  ↓
Better Prediction
```

### Important Idea

```text
Smaller Loss = Prediction is closer to the actual answer
Larger Loss  = Prediction is farther from the actual answer
```

The model's goal is generally to **reduce the loss during training**.

---

# 2. Loss Functions Based on Problem Type

Different problems need different loss functions.

| Problem                   | Common Loss          |
| ------------------------- | -------------------- |
| Regression                | MAE, MSE, RMSE       |
| Binary Classification     | Binary Cross Entropy |
| Multiclass Classification | Cross Entropy Loss   |

Example:

```text
Predict house price
        ↓
Regression
        ↓
MSE / MAE / RMSE
```

```text
Spam or Not Spam
        ↓
Binary Classification
        ↓
Binary Cross Entropy
```

```text
Cat / Dog / Horse
        ↓
Multiclass Classification
        ↓
Cross Entropy Loss
```

# 3. MAE – Mean Absolute Error

## Simple Concept

**MAE measures the average absolute difference between actual and predicted values.**

In simple words:

> How far is my prediction from the actual answer?

It does not care whether the prediction is above or below the actual value.

---

## Formula

```text
MAE = (1/n) × Σ |Actual - Prediction|
```

---

## Mathematical Intuition

Suppose we have:

```text
Actual      Prediction
50          45
60          65
70          68
```

Errors:

```text
|50 - 45| = 5
|60 - 65| = 5
|70 - 68| = 2
```

Average:

```text
MAE = (5 + 5 + 2) / 3

MAE = 4
```

So:

```text
MAE = 4
```

This means the model is **4 units away from the actual value on average**.

---

## Python Implementation

```python
import numpy as np

actual = np.array([50, 60, 70])
prediction = np.array([45, 65, 68])

mae = np.mean(np.abs(actual - prediction))

print("MAE:", mae)
```

### Output

```text
MAE: 4.0
```

---

## PyTorch Implementation

```python
import torch

actual = torch.tensor([50.0, 60.0, 70.0])
prediction = torch.tensor([45.0, 65.0, 68.0])

loss_function = torch.nn.L1Loss()

loss = loss_function(prediction, actual)

print("MAE:", loss.item())
```

### Output

```text
MAE: 4.0
```

`L1Loss()` is commonly used for MAE.

---

## Interpretation

```text
MAE = 4
```

means:

> The prediction is off by approximately 4 units on average.

Lower MAE is better.

---

## When to Use

Use MAE when:

* You want an easy-to-understand error.
* You don't want large errors to dominate too strongly.
* Your dataset may contain some outliers.

### Real-World Example

House price prediction:

```text
Actual:     ₹50 Lakhs
Prediction: ₹45 Lakhs
```

MAE focuses on the absolute ₹5 Lakh difference.

---

## When Not to Use

MAE may not be ideal when:

* Large errors should be punished heavily.
* You specifically want the model to strongly avoid large mistakes.

For those situations, MSE can be useful.

---

## Advantages

* Easy to understand.
* Same unit as the target.
* Less sensitive to outliers than MSE.

---

## Limitations

* Does not punish large errors as strongly as MSE.
* The absolute-value operation makes optimization less smooth around zero.

---

# 4. MSE – Mean Squared Error

## Simple Concept

MSE calculates the error and **squares it** before taking the average.

This means large errors become much more important.

---

## Formula

```text
MSE = (1/n) × Σ(Actual - Prediction)²
```

---

## Mathematical Intuition

Using the same example:

```text
Actual      Prediction

50          45
60          65
70          68
```

Errors:

```text
5
-5
2
```

Square them:

```text
5²  = 25
(-5)² = 25
2²  = 4
```

Average:

```text
MSE = (25 + 25 + 4) / 3

MSE = 18
```

So:

```text
MSE = 18
```

Notice something important:

```text
MAE = 4
MSE = 18
```

They measure error differently.

---

## Python Implementation

```python
import numpy as np

actual = np.array([50, 60, 70])
prediction = np.array([45, 65, 68])

mse = np.mean((actual - prediction) ** 2)

print("MSE:", mse)
```

### Output

```text
MSE: 18.0
```

---

## PyTorch Implementation

```python
import torch

actual = torch.tensor([50.0, 60.0, 70.0])
prediction = torch.tensor([45.0, 65.0, 68.0])

loss_function = torch.nn.MSELoss()

loss = loss_function(prediction, actual)

print("MSE:", loss.item())
```

### Output

```text
MSE: 18.0
```

---

## Interpretation

```text
MSE = 18
```

Lower MSE means predictions are generally closer to actual values.

But MSE is in **squared units**, so it is not as directly understandable as MAE.

---

## Why Does MSE Punish Large Errors?

Consider two errors:

```text
Error = 2
Error = 10
```

MAE:

```text
2
10
```

MSE:

```text
2²  = 4
10² = 100
```

The error of 10 becomes **100**.

Therefore:

> MSE gives much more importance to large mistakes.

---

## When to Use

Use MSE when:

* Large prediction errors are costly.
* You want the model to strongly avoid large mistakes.
* You are training many neural-network regression models.

### Real-World Example

Electricity demand prediction:

```text
Actual demand = 1000 MW
Prediction     = 1500 MW
```

A large error may be very important, so MSE can strongly penalize it.

---

## When Not to Use

Avoid relying on MSE when:

* Your data contains many extreme outliers.
* Large errors should not dominate the training process.

---

## Advantages

* Strongly penalizes large errors.
* Smooth and convenient for optimization.
* Very commonly used in regression.

---

## Limitations

* Sensitive to outliers.
* Squared units make interpretation less intuitive.

---

# 5. RMSE – Root Mean Squared Error

## Simple Concept

RMSE is basically:

```text
MSE → Take square root
```

---

## Formula

```text
RMSE = √MSE
```

or:

```text
RMSE = √[(1/n) × Σ(Actual - Prediction)²]
```

---

## Mathematical Intuition

From the previous example:

```text
MSE = 18
```

Therefore:

```text
RMSE = √18

RMSE ≈ 4.24
```

So:

```text
RMSE ≈ 4.24
```

---

## Python Implementation

```python
import numpy as np

actual = np.array([50, 60, 70])
prediction = np.array([45, 65, 68])

mse = np.mean((actual - prediction) ** 2)
rmse = np.sqrt(mse)

print("RMSE:", rmse)
```

### Output

```text
RMSE: 4.242640687119285
```

---

## PyTorch Implementation

```python
import torch

actual = torch.tensor([50.0, 60.0, 70.0])
prediction = torch.tensor([45.0, 65.0, 68.0])

mse_loss = torch.nn.MSELoss()

mse = mse_loss(prediction, actual)
rmse = torch.sqrt(mse)

print("RMSE:", rmse.item())
```

### Output

```text
RMSE: 4.242640495300293
```

---

## Interpretation

If:

```text
RMSE = 4.24
```

the model's prediction error is roughly **4.24 units**, with larger errors receiving more influence.

Unlike MSE, RMSE is in the **same unit as the original target**.

For example:

```text
House price → ₹ Lakhs
RMSE → ₹ Lakhs
```

---

## When to Use

Use RMSE when:

* You want large errors to matter.
* You want the result in the same unit as the target.
* You want an easy-to-communicate regression metric.

---

## When Not to Use

RMSE may not be suitable when:

* Your dataset contains many extreme outliers.
* You don't want large errors to have extra influence.

---

## Advantages

* Same unit as target.
* Penalizes large errors.
* Easy to communicate.

---

## Limitations

* Sensitive to outliers.
* Large errors can strongly affect the result.

---

# 6. MAE vs MSE vs RMSE

| Loss | Main Idea      | Outlier Effect |
| ---- | -------------- | -------------- |
| MAE  | Absolute error | Lower          |
| MSE  | Squared error  | High           |
| RMSE | √MSE           | High           |

### Easy Memory Trick

```text
MAE → How far?

MSE → How far, but punish big mistakes strongly?

RMSE → Same as MSE, but bring it back to original units.
```

# 7. Binary Cross Entropy

## Simple Concept

Binary classification has only **two possible classes**.

Examples:

```text
Spam / Not Spam
Fraud / Not Fraud
Disease / No Disease
Pass / Fail
```

Usually:

```text
0 = Negative
1 = Positive
```

The model produces a probability between:

```text
0 and 1
```

Example:

```text
Prediction = 0.90
```

means the model is saying:

> I am giving a 90% probability to class 1.

---

## Why Do We Need Binary Cross Entropy?

Suppose the actual answer is:

```text
1
```

Prediction:

```text
0.90
```

This is a good prediction.

But:

```text
Prediction = 0.10
```

is a bad prediction because the model was very confident in the wrong direction.

BCE gives:

* **Small loss** → confident and correct
* **Large loss** → confident and wrong

---

## Formula

For one example:

```text
BCE = -[y log(p) + (1-y) log(1-p)]
```

Where:

```text
y = actual label
p = predicted probability
```

---

## Simple Numerical Example

Actual:

```text
y = 1
```

Prediction:

```text
p = 0.9
```

Because `y = 1`, the formula becomes:

```text
BCE = -log(0.9)
```

Approximately:

```text
BCE ≈ 0.105
```

Small loss → good prediction.

Now suppose:

```text
y = 1
p = 0.1
```

Then:

```text
BCE = -log(0.1)
```

Approximately:

```text
BCE ≈ 2.303
```

Large loss → bad prediction.

### Important

```text
Actual = 1
Prediction = 0.9 → small loss

Actual = 1
Prediction = 0.1 → large loss
```

---

## Python Implementation

```python
import numpy as np

actual = 1
prediction = 0.9

bce = -(actual * np.log(prediction) +
        (1 - actual) * np.log(1 - prediction))

print("BCE:", bce)
```

### Output

```text
BCE: 0.10536051565782628
```

---

## PyTorch Implementation

```python
import torch

actual = torch.tensor([1.0])
prediction = torch.tensor([0.9])

loss_function = torch.nn.BCELoss()

loss = loss_function(prediction, actual)

print("BCE:", loss.item())
```

### Output

```text
BCE: 0.10536054521799088
```

---

## Interpretation

```text
BCE ≈ 0.105
```

This is a small loss, meaning the prediction was close to the correct class.

Remember:

> For loss functions, generally **lower is better**.

---

## When to Use

Use Binary Cross Entropy when:

* There are exactly two classes.
* Your model output is a probability between 0 and 1.

Examples:

```text
Spam / Not Spam
Fraud / Not Fraud
Customer Churn / No Churn
```

---

## When Not to Use

Don't use ordinary BCE for:

* Three or more mutually exclusive classes.

For example:

```text
Cat / Dog / Horse
```

Use Cross Entropy Loss instead.

---

## Advantages

* Designed for binary classification.
* Works naturally with probabilities.
* Strongly penalizes confident wrong predictions.

---

## Limitations

* Requires appropriate probability output.
* Can be numerically problematic if implemented manually without proper safeguards.

---

# 8. BCE With Logits – Important Concept

## Simple Concept

There are two related approaches.

### Approach 1

```text
Model
 ↓
Sigmoid
 ↓
Probability
 ↓
BCELoss
```

### Approach 2

```text
Model
 ↓
Raw score (logit)
 ↓
BCEWithLogitsLoss
```

With PyTorch, the second approach is usually preferred when your model produces raw logits.

---

## What Is a Logit?

A **logit is the raw output of the model before sigmoid**.

Example:

```text
Raw model output = 2.0
```

Sigmoid converts it to probability:

```text
Sigmoid(2.0) ≈ 0.88
```

So:

```text
Logit → Sigmoid → Probability
2.0   →          → 0.88
```

---

## PyTorch Example

```python
import torch

actual = torch.tensor([1.0])
logit = torch.tensor([2.0])

loss_function = torch.nn.BCEWithLogitsLoss()

loss = loss_function(logit, actual)

print("Loss:", loss.item())
```

The important point is:

```text
BCEWithLogitsLoss
```

already combines:

```text
Sigmoid + Binary Cross Entropy
```

---

## When to Use BCEWithLogitsLoss

Use it when your model outputs **raw logits**.

Example:

```python
model = torch.nn.Linear(4, 1)
```

The output of this layer is a raw score, not a probability.

Therefore:

```python
loss_function = torch.nn.BCEWithLogitsLoss()
```

is appropriate.

### Important

Do **not** do this unnecessarily:

```text
Linear
 ↓
Sigmoid
 ↓
BCEWithLogitsLoss
```

because `BCEWithLogitsLoss` already handles the sigmoid internally.

# 9. Cross Entropy Loss

## Simple Concept

Cross Entropy Loss is commonly used for **multiclass classification**.

Example:

```text
Image
 ↓
Model
 ↓
Cat / Dog / Horse
```

There are three possible classes.

The model produces scores for all classes.

Example:

```text
Cat   = 2.5
Dog   = 0.5
Horse = 1.0
```

Cross Entropy uses these scores to calculate how well the model selected the correct class.

---

## Important Point

PyTorch's `CrossEntropyLoss` expects **raw logits**, not probabilities.

Example:

```text
Cat   → 2.5
Dog   → 0.5
Horse → 1.0
```

These are logits.

PyTorch internally handles the necessary normalization.

---

## Formula

For the correct class:

```text
Cross Entropy = -log(probability of correct class)
```

---

## Simple Numerical Example

Suppose the actual class is:

```text
Dog
```

The model probabilities are:

```text
Cat   = 0.20
Dog   = 0.70
Horse = 0.10
```

Correct-class probability:

```text
Dog = 0.70
```

Therefore:

```text
Loss = -log(0.70)
```

Approximately:

```text
Loss ≈ 0.357
```

This is relatively small because the model gave the correct class a high probability.

---

## What If the Model Is Wrong?

Actual:

```text
Dog
```

Prediction:

```text
Cat = 0.90
Dog = 0.05
Horse = 0.05
```

Correct-class probability:

```text
Dog = 0.05
```

Therefore:

```text
Loss = -log(0.05)
```

Approximately:

```text
Loss ≈ 2.996
```

Large loss → the model was confidently wrong.

---

## Python Implementation

```python
import numpy as np

actual_class_probability = 0.70

cross_entropy = -np.log(actual_class_probability)

print("Cross Entropy Loss:", cross_entropy)
```

### Output

```text
Cross Entropy Loss: 0.35667494393873245
```

---

## PyTorch Implementation

```python
import torch

# Raw model outputs
logits = torch.tensor([[2.5, 0.5, 1.0]])

# Actual class:
# 0 = Cat
# 1 = Dog
# 2 = Horse
actual = torch.tensor([1])

loss_function = torch.nn.CrossEntropyLoss()

loss = loss_function(logits, actual)

print("Cross Entropy Loss:", loss.item())
```

### Output

Approximately:

```text
Cross Entropy Loss: 2.4076
```

Why is this different from the earlier `0.357`?

Because PyTorch first converts the **logits**:

```text
[2.5, 0.5, 1.0]
```

into probabilities.

The probability of Dog is not exactly `0.70`.

The important concept is:

> CrossEntropyLoss works directly with logits and calculates the loss for the correct class.

---

## When to Use

Use Cross Entropy Loss when:

* You have multiple mutually exclusive classes.

Examples:

```text
Cat / Dog / Horse

Low / Medium / High

Apple / Banana / Orange
```

---

## When Not to Use

For a simple two-class problem, Binary Cross Entropy is commonly used.

Also, if multiple labels can independently be true at the same time, a different setup such as binary classification per label is generally appropriate.

Example:

```text
Image can contain:
Dog = 1
Car = 1
Person = 1
```

This is **multilabel classification**, not ordinary mutually exclusive multiclass classification.

---

## Advantages

* Designed for multiclass classification.
* Works naturally with neural-network logits.
* Strongly penalizes confident wrong predictions.
* Very commonly used in Deep Learning.

---

## Limitations

* Not designed for ordinary regression.
* For multilabel problems, a different loss setup is generally needed.
* Class imbalance may require additional techniques.

# 11. Prediction → Loss → Learning

This is one of the **most important concepts in Deep Learning**.

Suppose we are training a neural network to detect spam.

### Step 1 – Input

```text
Email text
```

### Step 2 – Neural Network

The neural network processes the input.

```text
Email
 ↓
Neural Network
```

### Step 3 – Prediction

The model predicts:

```text
Spam probability = 0.30
```

### Step 4 – Actual Answer

Suppose the email is actually spam:

```text
Actual = 1
```

The prediction is not very good.

### Step 5 – Loss

BCE calculates the error.

```text
Prediction → 0.30
Actual     → 1
Loss       → Large
```

### Step 6 – Learning

The training process uses the loss to determine how the model's parameters should change.

```text
Loss
 ↓
Gradients
 ↓
Weights updated
 ↓
Model learns
```

### Step 7 – Next Prediction

After training, the model may produce:

```text
Spam probability = 0.75
```

The prediction is now closer to the actual answer.

# 12. Very Important Difference: Loss vs Metric

Do not confuse these two.

### Loss

Used during training to tell the model:

> How wrong are you?

Examples:

```text
MSE
BCE
Cross Entropy
```

### Metric

Used mainly to understand model performance.

Examples:

```text
Accuracy
Precision
Recall
F1-score
MAE
RMSE
```

Some measures, such as MAE or RMSE, can be used as evaluation metrics, while other losses are primarily designed for optimization.
