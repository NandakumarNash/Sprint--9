## 1. `nn.Module`

### Simple Concept

`nn.Module` is the base class used to create a neural network in PyTorch.

### Important Explanation

* It helps PyTorch manage layers, weights, biases, and parameters.
* Our neural network class inherits from `nn.Module`.
* Every custom PyTorch model normally starts with `nn.Module`.

### Python Implementation

```python
import torch
import torch.nn as nn

class CustomerPurchaseModel(nn.Module):

    def __init__(self):
        super().__init__()

model = CustomerPurchaseModel()

print(model)
```

### Output

```text
CustomerPurchaseModel()
```

### What Example Tells

We have created a basic PyTorch neural network model, but we haven't added any layers yet.

### Real-World Example

For customer purchase prediction:

```text
Customer Data
      ↓
Neural Network
      ↓
Purchase / No Purchase
```

The entire neural network can be created inside our `nn.Module` class.

### Business Example

An e-commerce company can use a neural network to predict whether a customer is likely to purchase a product.

### AI/ML Use Case

`nn.Module` is used whenever we build a custom neural network using PyTorch.

### Key Points

* `nn.Module` = base class for PyTorch models.
* Our model inherits from `nn.Module`.
* `super().__init__()` initializes PyTorch's model functionality.
* Layers and parameters will be added to this model.

## 2. Constructor (`__init__`)

### Simple Concept

The constructor is where we **define the layers of our neural network**.

### Important Explanation

The constructor runs when we create the model:

```python
model = CustomerPurchaseModel()
```

Inside `__init__()`, we define things like:

* Input layer
* Hidden layers
* Activation layers
* Output layer

### Python Implementation

```python
class CustomerPurchaseModel(nn.Module):

    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(4, 8)
        self.layer2 = nn.Linear(8, 4)
        self.output = nn.Linear(4, 1)
```

### What Example Tells

Here:

```text
4 → 8 → 4 → 1
```

means:

* 4 input features
* 8 neurons in first hidden layer
* 4 neurons in second hidden layer
* 1 output neuron

### Key Points

* `__init__()` defines the model structure.
* Layers are usually created inside `__init__()`.
* `self.layer1`, `self.layer2`, etc. store the layers inside the model.

## 3. Layers in a Neural Network

### Simple Concept

A layer contains neurons that process information and pass it to the next layer.

### Example

```text
Customer Data
     ↓
Input: 4 features
     ↓
Hidden Layer: 8 neurons
     ↓
Hidden Layer: 4 neurons
     ↓
Output: 1 neuron
```

### Key Points

* Input features are the information given to the model.
* Hidden neurons learn patterns from the data.
* Output neurons produce the final prediction.

## 5. Forward Method

### Simple Concept

The `forward()` method defines **how data moves through the neural network**.

### Python Example

```python
def forward(self, x):

    x = self.layer1(x)
    x = self.layer2(x)
    x = self.output(x)

    return x
```

### Data Flow

```text
Input
  ↓
layer1
  ↓
layer2
  ↓
output
  ↓
Prediction
```

### Key Points

* `forward()` controls the data flow.
* It tells PyTorch which layer should process the data first.
* It returns the model's output.

---

## 6. Activation Layers

### Simple Concept

Activation functions help the neural network learn **non-linear patterns**.

A common activation function is ReLU.

### Python Example

```python
self.relu = nn.ReLU()
```

Then inside `forward()`:

```python
x = self.layer1(x)
x = self.relu(x)
```

### Data Flow

```text
Input
  ↓
Linear Layer
  ↓
ReLU
  ↓
Next Layer
```

### Key Points

* ReLU is commonly used between hidden layers.
* It helps the network learn complex patterns.
* The output layer may use a different activation depending on the problem.

## 8. Complete Neural Network Architecture

### Simple Concept

Now we combine everything into one neural network.

### Python Implementation

```python
import torch
import torch.nn as nn

class CustomerPurchaseModel(nn.Module):

    def __init__(self):
        super().__init__()

        self.layer1 = nn.Linear(4, 8)
        self.relu1 = nn.ReLU()

        self.layer2 = nn.Linear(8, 4)
        self.relu2 = nn.ReLU()

        self.output = nn.Linear(4, 1)

    def forward(self, x):

        x = self.layer1(x)
        x = self.relu1(x)

        x = self.layer2(x)
        x = self.relu2(x)

        x = self.output(x)

        return x


model = CustomerPurchaseModel()

print(model)
```

### Architecture

```text
4 Input Features
       ↓
8 Hidden Neurons
       ↓
ReLU
       ↓
4 Hidden Neurons
       ↓
ReLU
       ↓
1 Output Neuron
```

## 9. Model Parameters & Trainable Parameters

### Simple Concept

Parameters are the **weights and biases that the neural network learns during training**.

### Python Implementation

```python
total_parameters = sum(p.numel() for p in model.parameters())

trainable_parameters = sum(
    p.numel() for p in model.parameters()
    if p.requires_grad
)

print("Total Parameters:", total_parameters)
print("Trainable Parameters:", trainable_parameters)
```

### What Example Tells

For this model, PyTorch counts all the weights and biases automatically.

If:

```text
requires_grad = True
```

the parameter can be learned during training.

### Key Points

* Parameters include weights and biases.
* Trainable parameters are parameters PyTorch updates during training.
* More neurons/layers usually means more parameters.

## 10. Inspect Weight Matrices & Model Architecture

### Simple Concept

We can inspect the actual weights learned by each `nn.Linear` layer.

### Python Implementation

```python
print(model)
```

To inspect the first layer:

```python
print(model.layer1.weight)
```

To inspect its bias:

```python
print(model.layer1.bias)
```

We can also inspect all parameters:

```python
for name, parameter in model.named_parameters():
    print(name)
    print(parameter.shape)
    print()
```

```text
4 input features
8 neurons
```

```text
4 × 8
```

```text
4 → 8 → 4 → 1
```
