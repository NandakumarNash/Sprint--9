# 1. What is a Gradient?

### Simple Concept

A **gradient tells us how much the loss changes when a parameter changes**.

In simple words:

> **Gradient tells the model which direction and how strongly a weight should change to reduce the error.**

For example:

* Gradient = `+5` → weight needs to move downward
* Gradient = `-5` → weight needs to move upward
* Gradient ≈ `0` → changing the weight has very little effect on the loss

### Real-World Example

Imagine a food-delivery model predicts:

**Delivery Time = 30 minutes**

Actual delivery time:

**40 minutes**

The model is wrong by 10 minutes.

The model needs to adjust its weights so that its future prediction becomes closer to 40.

The **gradient tells the model how the weight contributed to this error and which direction to adjust it.**

---

# 2. Why are Gradients Required?

A neural network has many **weights**.

During training, the model needs to answer:

> "How should I change each weight to reduce the loss?"

The gradient gives this information.

Without gradients, the model would not know how to efficiently adjust its weights.

The basic idea is:

```text
Prediction
    ↓
Calculate Loss
    ↓
Calculate Gradients
    ↓
Change Weights
    ↓
Better Prediction
```

---

# 3. Gradient Descent

### Simple Concept

**Gradient Descent** is the method used to update weights so that the loss becomes smaller.

The basic idea:

```text
Current weight
      ↓
Check gradient
      ↓
Move weight in the correct direction
      ↓
Calculate loss again
      ↓
Repeat
```

The update idea is:

```text
new weight = old weight - learning rate × gradient
```

The **learning rate** controls how big the weight update is.

For example:

```text
old weight = 5
gradient = 2
learning rate = 0.1
```

Then:

```text
new weight = 5 - (0.1 × 2)
           = 4.8
```

The weight moved from `5` to `4.8` because the gradient was positive.

---

# 4. Computational Graph

### Simple Concept

A **computational graph** shows the mathematical operations used to produce the final output.

Example:

```text
x
↓
multiply by w
↓
z = x × w
↓
calculate loss
↓
L
```

For example:

```text
x = 2
w = 3

z = x × w
  = 2 × 3
  = 6
```

Then suppose the target is `10`.

```text
Prediction = 6
Target     = 10
```

The loss tells us how wrong the prediction is.

The computational graph helps us trace **how each value contributed to the final loss**.

---

# 5. Forward Pass

### Simple Concept

The **forward pass** means:

> Input goes through the network → prediction is produced.

Example:

```text
Input
  ↓
Weight
  ↓
Prediction
```

Suppose:

```text
x = 2
w = 3
```

Then:

```text
prediction = x × w
           = 2 × 3
           = 6
```

This is the **forward pass**.

---

# 6. Loss Calculation

After the forward pass, we compare:

```text
Prediction vs Actual Target
```

Suppose:

```text
Prediction = 6
Target = 10
```

Using squared error:

```text
Loss = (prediction - target)²
     = (6 - 10)²
     = (-4)²
     = 16
```

So:

```text
Loss = 16
```

The model needs to reduce this loss.

---

# 7. Backward Pass

### Simple Concept

The **backward pass** works backward from the loss.

It asks:

> "How much did each weight contribute to this loss?"

The answer is represented by **gradients**.

```text
Forward:

x → weight → prediction → loss
                         ↓
Backward:               gradient
                         ↓
                       weight
```

So:

```text
Forward Pass
      ↓
Loss
      ↓
Backward Pass
      ↓
Gradients
```

---

# 8. Chain Rule

### Simple Concept

The **chain rule** allows us to calculate how one value affects another value through multiple operations.

Consider:

```text
x → w → prediction → loss
```

The weight affects the prediction.

The prediction affects the loss.

Therefore, we can calculate:

```text
How weight affects loss
```

by multiplying the effects along the path.

This is the basic mathematical idea behind **backpropagation**.

# 9. Small Manual Gradient Example

Let's use:

```text
x = 2
w = 3
target = 10
```

Our model:

```text
prediction = x × w
```

Therefore:

```text
prediction = 2 × 3
           = 6
```

Loss:

```text
Loss = (prediction - target)²

     = (6 - 10)²

     = 16
```

Now we want:

```text
dLoss / dw
```

That means:

> How much does the loss change when the weight changes?

Because:

```text
prediction = x × w
```

we get:

```text
dPrediction/dw = x = 2
```

And:

```text
Loss = (prediction - target)²
```

so:

```text
dLoss/dPrediction = 2 × (prediction - target)

                   = 2 × (6 - 10)

                   = -8
```

Using the chain rule:

```text
dLoss/dw
=
dLoss/dPrediction × dPrediction/dw

=
(-8) × 2

=
-16
```

Therefore:

```text
Gradient = -16
```

### What does `-16` mean?

It tells us that increasing the weight would reduce the loss in this situation.

The **negative sign tells the direction**.

The magnitude `16` tells us the loss is quite sensitive to this weight.

---

# 10. Weight Update

Now use gradient descent.

Suppose:

```text
old weight = 3
gradient = -16
learning rate = 0.1
```

Update:

```text
new weight = old weight - learning rate × gradient

           = 3 - (0.1 × -16)

           = 3 + 1.6

           = 4.6
```

So:

```text
Old weight = 3
New weight = 4.6
```

Why did the weight increase?

Because the gradient was negative.

```text
Gradient = -16
          ↓
Weight needs to increase
```

This is the core idea of **learning** in a neural network.

# 11. Backpropagation

### Simple Concept

**Backpropagation** is the process of calculating gradients by moving backward through the computational graph.

The complete process is:

```text
Input
  ↓
Forward Pass
  ↓
Prediction
  ↓
Calculate Loss
  ↓
Backward Pass
  ↓
Calculate Gradients
  ↓
Update Weights
  ↓
Repeat
```

This happens again and again during neural-network training.

---

# 12. PyTorch – `requires_grad=True`

PyTorch can automatically track calculations involving tensors.

We tell PyTorch:

```python
requires_grad=True
```

when we want PyTorch to calculate the gradient of that tensor.

### Python Example

```python
import torch

w = torch.tensor(3.0, requires_grad=True)

print(w)
```

### Output

```text
tensor(3., requires_grad=True)
```

This means:

> "PyTorch, keep track of calculations involving `w` because I may need its gradient."

---

# 13. PyTorch – `.backward()`

Now let's perform the same mathematical example using PyTorch.

```python
import torch

x = torch.tensor(2.0)

w = torch.tensor(3.0, requires_grad=True)

target = torch.tensor(10.0)

prediction = x * w

loss = (prediction - target) ** 2

loss.backward()

print("Prediction:", prediction.item())
print("Loss:", loss.item())
print("Gradient:", w.grad.item())
```

### Output

```text
Prediction: 6.0
Loss: 16.0
Gradient: -16.0
```

Notice:

```text
w.grad = -16
```

This is exactly the gradient we calculated manually.

---

# 14. What does `.backward()` actually do?

When we write:

```python
loss.backward()
```

PyTorch goes backward through the computational graph and calculates the gradients.

In our example:

```text
x = 2
w = 3
   ↓
prediction = 6
   ↓
loss = 16
   ↓
backward()
   ↓
gradient of w = -16
```

So `.backward()` basically tells PyTorch:

> **"Start from this loss and calculate how the loss changes with respect to the parameters."**

---

# 15. What is `.grad`?

After:

```python
loss.backward()
```

we can check:

```python
w.grad
```

This contains the gradient calculated for `w`.

Example:

```python
print(w.grad)
```

Output:

```text
tensor(-16.)
```

So remember:

```text
requires_grad=True
        ↓
Track the tensor

.backward()
        ↓
Calculate gradients

.grad
        ↓
Read the calculated gradient
```

---

# 16. How PyTorch Autograd Helps

### Simple Concept

**Autograd** is PyTorch's automatic differentiation system.

It automatically:

1. Tracks mathematical operations.
2. Builds the computational graph.
3. Calculates gradients when `.backward()` is called.
4. Stores the gradient in `.grad`.

Without autograd, we would have to manually calculate gradients for every weight.

A neural network may contain **thousands or millions of parameters**, so doing this manually would be extremely difficult.

Autograd does the calculation for us.

# 17. Complete Learning Cycle

The most important thing to remember from this notebook is:

```text
        INPUT
          ↓
    Forward Pass
          ↓
     Prediction
          ↓
    Calculate Loss
          ↓
    Backward Pass
          ↓
 Calculate Gradients
          ↓
    Update Weights
          ↓
     Repeat Again
```

In PyTorch:

```text
requires_grad=True
        ↓
Track calculations
        ↓
loss.backward()
        ↓
Calculate gradients
        ↓
w.grad
        ↓
Gradient available
        ↓
Optimizer updates weight
```

### One-line memory trick

> **Forward = make prediction.
> Loss = measure mistake.
> Backward = find how weights caused the mistake.
> Gradient = direction/strength of change.
> Update = change the weights.**
